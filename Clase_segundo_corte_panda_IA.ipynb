# Introducción a Pandas

## 1. ¿Qué es Pandas?

Pandas es una librería de Python muy usada para análisis y manipulación de datos.

El nombre viene de Panel Data, un término en estadística.

Está construida sobre NumPy, lo que le da velocidad y soporte para operaciones matemáticas.

👉 En palabras simples:
# Pandas nos permite  (leer, organizar, limpiar, transformar y analizar datos)

## 2. ¿Cuándo usamos Pandas?

Usamos Pandas principalmente cuando necesitamos:

- Leer datos de archivos como **CSV, Excel, SQL, JSON, etc.**

- Limpiar datos **(valores faltantes, duplicados, formatos).**

- **Seleccionar y filtrar filas y columnas.**

- **Transformar datos (crear nuevas columnas, aplicar funciones).**

- **Analizar datos (estadísticas, agrupaciones, conteo, promedios).**

- **Visualizar datos en conjunto con librerías como Matplotlib o Seaborn.**
