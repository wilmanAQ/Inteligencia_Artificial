# Visualización de la distribución normal de datos aleatorios

En este notebook generaremos un conjunto de datos aleatorios con distribución normal estándar utilizando NumPy y visualizaremos su distribución mediante un histograma y la curva teórica de la normal.

## 1. Generar datos aleatorios con distribución normal

Utilizaremos la función `np.random.randn(50)` para generar 50 datos aleatorios con distribución normal estándar (media 0 y desviación estándar 1). Mostraremos los datos generados y calcularemos su rango.

```python
import numpy as np

# Generar 50 datos aleatorios con distribución normal estándar
datos_rand = np.random.randn(50)
print("Datos generados:")
print(datos_rand)

# Calcular el rango de los datos
def rango_lista(datos):
    return max(datos) - min(datos)

print(f'El rango de datos_rand es: {rango_lista(datos_rand)}')
```

## 2. Graficar la distribución normal de los datos generados

A continuación, graficaremos un histograma de los datos generados y superpondremos la curva de densidad de la distribución normal teórica para comparar visualmente ambas distribuciones.

```python
import matplotlib.pyplot as plt
from scipy.stats import norm
import numpy as np

# Histograma de los datos generados
plt.figure(figsize=(8,5))
count, bins, ignored = plt.hist(datos_rand, bins=10, density=True, alpha=0.6, color='skyblue', edgecolor='black', label='Datos generados')

# Curva de densidad normal teórica
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, np.mean(datos_rand), np.std(datos_rand))
plt.plot(x, p, 'r--', linewidth=2, label='Curva normal teórica')

plt.title('Histograma de datos aleatorios y curva de distribución normal')
plt.xlabel('Valor')
plt.ylabel('Densidad')
plt.legend()
plt.grid(True)
plt.show()
```
